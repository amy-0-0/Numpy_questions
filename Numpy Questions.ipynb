## 3. Analyze a stock portfolio using NumPy's mathematical functions:
np.random.seed(789)
days = 252  # Trading days in a year
stocks = 5

Simulate daily returns (percentage changes)
daily_returns = np.random.normal(0.001, 0.02, (days, stocks))  # Mean 0.1% daily return, 2% volatility
stock_names = ['AAPL', 'GOOGL', 'MSFT', 'TSLA', 'AMZN']
initial_prices = np.array([150, 2800, 300, 800, 3200])

Tasks :
1. Basic Analysis
 -Calculate cumulative returns for each stock
 -Find annualized volatility (standard deviation * sqrt(252))
 -Identify best and worst performing stocks
2. Risk Metrics
 -Calculate Sharpe ratio (return/volatility) for each stock
 -Find maximum drawdown (largest peak-to-trough decline)
 -Calculate correlation matrix between stocks

```python
np.random.seed(789)
```

```python
days = 252  # Trading days in a year
stocks = 5
```

```python
#Simulate daily returns (percentage changes)
daily_returns = np.random.normal(0.001, 0.02, (days, stocks))  # Mean 0.1% daily return, 2% volatility
stock_names = ['AAPL', 'GOOGL', 'MSFT', 'TSLA', 'AMZN']
initial_prices = np.array([150, 2800, 300, 800, 3200])
```

```python
daily_returns
```

### Task 1

```python
# Basic analysis
## Calculate cumulative returns for each stock
cumulative_returns = np.prod(1 + daily_returns, axis=0) - 1
```

```python
cumulative_returns
```

```python
# Find annualized volatility (standard deviation * sqrt(252))
daily_std = np.std(daily_returns, axis=0)
```

```python
annualized_volatility = daily_std * np.sqrt(252)
```

```python
annualized_volatility
```

```python
# Best and worst performing stocks
best = np.argmax(cumulative_returns)
worst = np.argmin(cumulative_returns)
```

```python
best
```

```python
worst
```

### Task 2

```python
# Risk Metrics 
# Calculate Sharpe ratio (return/volatility) for each stock 
sharpe_ratio = cumulative_returns / annualized_volatility
```

```python
sharpe_ratio
```

```python
# Find maximum drawdown (largest peak-to-trough decline) 
prices = initial_prices * np.cumprod(1 + daily_returns, axis=0) #prices over time
peak = np.maximum.accumulate(prices, axis=0) #running peak
drawdown = (prices - peak) / peak #drawdown
max_drawdown = np.min(drawdown, axis=0) #maximum drawdown
```

```python
max_drawdown
```

```python
# Calculate correlation matrix between stocks
correlation_matrix = np.corrcoef(daily_returns, rowvar = False)
```
